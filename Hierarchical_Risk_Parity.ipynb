```python
"""
Hierarchical risk parity (HRP)

"""
!pip install pypfopt
import pandas as pd
import numpy as np
from pypfopt import HRPOpt

df_ret = pd.read_csv("downloaded_data.csv")

#Check data structure
print(df_ret.head())

#HRP optimisation
hrp = HRPOpt(df_ret)

#Returns an ordered dictionary of weights
hrp_weights = hrp.optimize(linkage_method="single")

#Convert to Pandas series
hrp_w = pd.Series(hrp_weights)
hrp_w = hrp_w.reindex(df_ret.columns).fillna(0)

print(hrp_w.sort_values(ascending=False))

```
