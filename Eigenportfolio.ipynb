```python
"""
Eigenportfolio
"""

import numpy as np
import pandas as pd

def eigenportfolio(df: pd.DataFrame):
    """
    Parameters
    ----------
    df:					DataFrame, returns.

    Returns
    ---------
    eigenvalues: 		np.ndarray, eigenvalues, descending order.
    eigenportfolios:	DataFrame, eigenvectors (weights) sorted by eigenvalue.
    """

    #Eigen decomposition
    eigenvalues, eigenvectors = np.linalg.eigh(df.corr())

    #Sort by eigenvalue
    idx = np.argsort(eigenvalues)[::-1]
    eigenvalues = eigenvalues[idx]
    eigenvectors = eigenvectors[:, idx]

    #Normalize eigenvectors so weights sum to 1
    eigenvectors = eigenvectors / np.sum(eigenvectors, axis=0)

    #Convert to DataFrame for readability
    eigenportfolios = pd.DataFrame(
        eigenvectors,
        index=df.columns,
        columns=[f"eigen_port_{i+1}" for i in range(len(eigenvalues))]
    )

    return eigenvalues, eigenportfolios
```

```python
#Implementation
df_ret = pd.read_csv("asset_classes_data.csv")

#Check data structure
print(df_ret.head())

eigenvalues, eigenportfolios = eigenportfolio(df_ret)

print("Eigenvalues:")
print(eigenvalues)
print("\nEigenportfolios:")
print(eigenportfolios)


```
